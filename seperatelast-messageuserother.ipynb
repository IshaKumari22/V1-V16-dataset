```python
# seperate the last message for user and other person

import json

# ==========================
# File Paths
# ==========================
input_file = "./dataset/finalV1B/v1b_training_dataset.jsonl"

user_output = "./dataset/finalV1B/v1b_last_user.jsonl"
other_output = "./dataset/finalV1B/v1b_last_other.jsonl"

user_count = 0
other_count = 0
error_count = 0

with open(input_file, "r", encoding="utf-8") as infile, \
     open(user_output, "w", encoding="utf-8") as user_file, \
     open(other_output, "w", encoding="utf-8") as other_file:

    for line_num, line in enumerate(infile, start=1):

        line = line.strip()

        if not line:
            continue

        try:
            sample = json.loads(line)

            conversation = sample["input"]["conversation"].strip()

            # Get all conversation lines
            conv_lines = [
                x.strip()
                for x in conversation.split("\n")
                if x.strip()
            ]

            if len(conv_lines) == 0:
                error_count += 1
                continue

            last_message = conv_lines[-1]

            if last_message.startswith("User:"):
                user_file.write(
                    json.dumps(sample, ensure_ascii=False) + "\n"
                )
                user_count += 1
            else:
                other_file.write(
                    json.dumps(sample, ensure_ascii=False) + "\n"
                )
                other_count += 1

        except Exception as e:
            print(f"Error on line {line_num}: {e}")
            error_count += 1

print("=" * 50)
print("Finished!")
print("=" * 50)
print(f"Last message = User : {user_count}")
print(f"Last message = Other: {other_count}")
print(f"Errors              : {error_count}")
print("=" * 50)
```
